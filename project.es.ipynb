#### Ejercicio 07. ¿Cuál es la media de precios en la población de "Arroyomolinos (Madrid)"? (★★☆)

Imprima el valor obtenido de la columna level5.

```python
nonull_ds = nonull_ds[nonull_ds['price'] != 0]
```

```python
# TODO
arroyomolinos = nonull_ds.filter(items=['level5', 'price'])
arroyomolinos = arroyomolinos[arroyomolinos['level5'] == 'Arroyomolinos (Madrid)']
print(arroyomolinos['price'].mean())
print(arroyomolinos['price'].min())
```

#### Ejercicio 08. Trazar el histograma de los precios para la población de "Arroyomolinos (Madrid)" y explica qué observas (★★☆)

Imprime el histograma de los precios y escribe en la celda del Markdown un breve análisis del trazado.


```python
import matplotlib.pyplot as plt

# Trazar el histograma de precios
plt.hist(arroyomolinos['price'], bins=20)
plt.xlabel('Precio')
plt.ylabel('Frecuencia')
plt.title('Histograma de Precios para Arroyomolinos (Madrid)')
plt.show()
```
